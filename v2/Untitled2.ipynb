```python
import numpy as np
from scipy.stats import norm
from bokeh.plotting import figure
from bokeh.plotting import figure, show, output_notebook
support_lower_bound = 0
support_upper_bound = 2.3
sigma_scalar=.15
support_discretization_factor = .001
support = tuple(np.arange(support_lower_bound, support_upper_bound, support_discretization_factor))
s = support[1000]

sigma = s * sigma_scalar
pdf = norm.pdf(support, loc=s, scale=sigma)
fig = figure(title='Pdf(X)', x_axis_label='Support', y_axis_label='Pdf(X)')

normal = norm.pdf(support, loc=0, scale=1)

fig.line(support, pdf, line_width=1, legend_label='Recomputed on fly')
fig.line(support, normal, line_width=1, legend_label='Recomputed on fly', color='pink')

output_notebook()
show(fig)
```

```python
def calculate_normal_pdf(support, means, sigmas):
    assert len(means) == len(sigmas), 'expects means & sigmas to be of equal sizes'
    support_size = len(support)
    size = len(means)
    total_size = support_size * size
    
    supports = np.tile(support, size).reshape((total_size,))
    repeated_means = np.repeat(means, support_size).reshape((total_size,))
    repeated_sigmas = np.repeat(sigmas, support_size).reshape((total_size,))
    
    normalization_constant = np.sqrt(2*np.pi)
    normalization_constants = np.repeat(1/(normalization_constant*np.array(sigmas)), support_size).reshape((total_size,))
    
    ys = (supports - repeated_means)/repeated_sigmas
    
    return (normalization_constants * np.exp(-(ys**2)/2)).reshape((size, support_size))   

# x  = np.arange(-5, 5, .02)
# x_pdfs = np.tile(x, 6).reshape((len(x)*6,))
# x_means = np.repeat((0, 1, 2, 3, 4, 5), len(x)).reshape((len(x)*6,))
# x_sigmas = np.repeat((1, 1.5, 2, 2.5, 3, 3.5), len(x)).reshape((len(x)*6,))
# normalization_constant = np.sqrt(2*np.pi)
# normalization_constants = np.repeat(1/(normalization_constant*np.array([1, 1.5, 2, 2.5, 3, 3.5])), len(x)).reshape((len(x)*6,))
# ys = (x_pdfs - x_means)/x_sigmas
# pdfs = (normalization_constants * np.exp(-(ys**2)/2)).reshape((6, len(x)))
support = np.arange(0, 2, .001)
pdfs = calculate_normal_pdf(support, np.arange(0, 3000), np.arange(1,3001))
```
